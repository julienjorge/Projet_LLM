```python
import pdfplumber

# Nom du fichier PDF à parser
PDF_PATH = "/Users/AI/Mes Documents/Docugami/KG-RAG-datasets/nih-clinical-trial-protocols/data/v1/docs/NCT06145295_Prot_SAP_000.pdf"

# Fichier texte où on va sauvegarder le contenu
OUTPUT_TEXT_PATH = "/Users/AI/Desktop/test_parsing_PDF.txt"

def extract_text_from_pdf(pdf_path, output_path):
    # Ouvre le PDF avec un contexte "with" (il sera bien fermé même en cas d'erreur)
    with pdfplumber.open(pdf_path) as pdf:
        # "pdf.pages" contient toutes les pages du document
        all_text = []

        for page_number, page in enumerate(pdf.pages, start=1):
            print(f"Lecture de la page {page_number}/{len(pdf.pages)}...")

            # Récupère le texte brut de la page
            text = page.extract_text()

            # Certaines pages peuvent ne rien retourner (text = None)
            if text:
                all_text.append(f"=== Page {page_number} ===\n{text}\n")
            else:
                all_text.append(f"=== Page {page_number} ===\n(Aucun texte trouvé)\n")

    # Sauvegarde tout le texte dans un fichier
    with open(output_path, "w", encoding="utf-8") as f:
        f.write("\n".join(all_text))

    print(f"Extraction terminée. Texte sauvegardé dans : {output_path}")


if __name__ == "__main__":
    extract_text_from_pdf(PDF_PATH, OUTPUT_TEXT_PATH)
```

```python
import pdfplumber
import os

# Dossier contenant les PDF
INPUT_FOLDER = "/Users/AI/Mes Documents/Docugami/KG-RAG-datasets/nih-clinical-trial-protocols/data/v1/docs"
# Dossier où seront enregistrés les fichiers texte
OUTPUT_FOLDER = "/Users/AI/Desktop/text_parsing"

def extract_text_from_pdf(pdf_path, output_path):
    with pdfplumber.open(pdf_path) as pdf:
        all_text = []

        for page_number, page in enumerate(pdf.pages, start=1):
            print(f"Lecture de {os.path.basename(pdf_path)} - page {page_number}/{len(pdf.pages)}...")

            text = page.extract_text()

            if text:
                all_text.append(f"=== Page {page_number} ===\n{text}\n")
            else:
                all_text.append(f"=== Page {page_number} ===\n(Aucun texte trouvé)\n")

    with open(output_path, "w", encoding="utf-8") as f:
        f.write("\n".join(all_text))

    print(f"➡ Extraction terminée pour {pdf_path} → {output_path}\n")


def process_all_pdfs(input_folder, output_folder):
    # Crée le dossier de sortie s'il n'existe pas
    os.makedirs(output_folder, exist_ok=True)

    # Liste tous les PDF du dossier
    pdf_files = [f for f in os.listdir(input_folder) if f.lower().endswith(".pdf")]

    if not pdf_files:
        print("Aucun fichier PDF trouvé dans le dossier.")
        return

    for pdf_file in pdf_files:
        pdf_path = os.path.join(input_folder, pdf_file)
        
        # Génère le nom du fichier .txt correspondant
        txt_filename = os.path.splitext(pdf_file)[0] + ".txt"
        output_path = os.path.join(output_folder, txt_filename)

        extract_text_from_pdf(pdf_path, output_path)


if __name__ == "__main__":
    process_all_pdfs(INPUT_FOLDER, OUTPUT_FOLDER)
```
