```python
import os
from pathlib import Path
from langchain_community.document_loaders import DirectoryLoader, TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

print("="*70)
print(" DIAGNOSTIC : Pourquoi seulement 5 documents ?")
print("="*70)

# ========================================
# ÉTAPE 1 : VÉRIFIER COMBIEN DE FICHIERS EXISTENT
# ========================================

folder_path = r"C:\Users\lehna\Downloads\text_output"

print(f"\n1️ Vérification du dossier : {folder_path}\n")

# Lister TOUS les fichiers .txt
all_txt_files = list(Path(folder_path).glob("**/*.txt"))
print(f"✓ Nombre total de fichiers .txt trouvés : {len(all_txt_files)}")

# Afficher les 10 premiers
print(f"\n Premiers fichiers trouvés :")
for i, file in enumerate(all_txt_files[:10], 1):
    file_size = os.path.getsize(file)
    print(f"   {i}. {file.name} ({file_size:,} bytes)")

if len(all_txt_files) > 10:
    print(f"   ... et {len(all_txt_files) - 10} autres fichiers")
```

```python
print(f"\n{'='*70}")
print("2️ Chargement des documents avec gestion d'erreurs...")
print('='*70)

docs = []
failed_files = []

for file_path in all_txt_files:
    try:
        # Essayer plusieurs encodages
        for encoding in ['utf-8', 'latin-1', 'cp1252', 'iso-8859-1']:
            try:
                loader = TextLoader(str(file_path), encoding=encoding)
                file_docs = loader.load()
                docs.extend(file_docs)
                print(f"✓ Chargé : {file_path.name} (encoding: {encoding})")
                break  # Si succès, sortir de la boucle
            except UnicodeDecodeError:
                continue  # Essayer l'encoding suivant
            except Exception as e:
                print(f"Erreur avec {file_path.name} : {e}")
                failed_files.append((file_path.name, str(e)))
                break
    except Exception as e:
        print(f"Impossible de charger {file_path.name} : {e}")
        failed_files.append((file_path.name, str(e)))

print(f"\n Résumé du chargement :")
print(f" Documents chargés : {len(docs)}")
print(f" Fichiers échoués : {len(failed_files)}")

if failed_files:
    print(f"\n Fichiers qui ont échoué :")
    for filename, error in failed_files:
        print(f"   - {filename} : {error[:50]}...")

# Vérifier si on a bien tous les documents
if len(docs) < len(all_txt_files):
    print(f"\n ATTENTION : {len(all_txt_files) - len(docs)} fichiers n'ont pas été chargés !")
```

```python
print(f"\n{'='*70}")
print("3️ MÉTHODE ALTERNATIVE : Chargement manuel robuste")
print('='*70)

def load_documents_robust(folder_path):
    """Charge tous les fichiers .txt avec gestion d'erreurs robuste"""
    from langchain_core.documents import Document
    
    documents = []
    all_files = list(Path(folder_path).glob("*.txt"))  # Fichiers directs
    all_files.extend(list(Path(folder_path).glob("**/*.txt")))  # Sous-dossiers
    all_files = list(set(all_files))  # Supprimer doublons
    
    print(f"Tentative de chargement de {len(all_files)} fichiers...\n")
    
    for file_path in all_files:
        try:
            # Essayer de lire avec différents encodings
            content = None
            for enc in ['utf-8', 'latin-1', 'cp1252', 'iso-8859-1']:
                try:
                    with open(file_path, 'r', encoding=enc) as f:
                        content = f.read()
                    break
                except:
                    continue
            
            if content and len(content.strip()) > 0:
                doc = Document(
                    page_content=content,
                    metadata={"source": str(file_path)}
                )
                documents.append(doc)
                print(f" {file_path.name} ({len(content):,} chars)")
            else:
                print(f" {file_path.name} est vide")
                
        except Exception as e:
            print(f" {file_path.name} : {str(e)[:50]}")
    
    return documents

# Charger avec la méthode robuste
docs_robust = load_documents_robust(folder_path)

print(f"\n Total chargé avec méthode robuste : {len(docs_robust)} documents")
```

```python
print(f"\n{'='*70}")
print("4️ Création de la vectorstore avec TOUS les documents")
print('='*70)

# Utiliser les documents chargés (prendre la meilleure méthode)
final_docs = docs_robust if len(docs_robust) > len(docs) else docs

print(f"\nNombre de documents à traiter : {len(final_docs)}")

# Afficher un exemple
if final_docs:
    print(f"\n Exemple de document chargé :")
    print(f"   Source : {final_docs[0].metadata.get('source', 'N/A')}")
    print(f"   Longueur : {len(final_docs[0].page_content):,} caractères")
    print(f"   Début : {final_docs[0].page_content[:200]}...")
```

```python
print(f"\n5️ Découpage en chunks...")

splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200,
    separators=["\n\n", "\n", ". ", " ", ""],
    length_function=len,
)

splitted_docs = splitter.split_documents(final_docs)
print(f"✓ {len(splitted_docs)} chunks créés")
```

```python
print(f"\n6️ Création de la vectorstore...")

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore_final = Chroma.from_documents(
    documents=splitted_docs,
    embedding=embeddings,
    persist_directory="chroma_db_COMPLETE"  # Nouveau nom !
)

vectorstore_final.persist()

total_stored = vectorstore_final._collection.count()
print(f"\n Vectorstore créée : 'chroma_db_COMPLETE'")
print(f" {total_stored} chunks stockés")
```

```python
print(f"\n{'='*70}")
print("7️ TEST DE QUALITÉ")
print('='*70)

query = "According to the protocol, what is the main objective of this clinical study?"
results = vectorstore_final.similarity_search_with_score(query, k=5)

print(f"\n🔍 Query : '{query}'\n")

for i, (doc, score) in enumerate(results, 1):
    status = "EXCELLENT" if score < 0.5 else ("⚠️ BON" if score < 0.8 else "❌ MAUVAIS")
    print(f"--- Résultat {i} ---")
    print(f"Score : {score:.4f} {status}")
    print(f"Source : {Path(doc.metadata.get('source', 'N/A')).name}")
    print(f"Extrait : {doc.page_content[:150]}...")
    print()
```

```python
import os
from dotenv import load_dotenv

load_dotenv()
MISTRAL_API_KEY = os.getenv("MISTRAL_API_KEY")
if not MISTRAL_API_KEY:
    raise ValueError("MISTRAL_API_KEY manquante. Mets-la dans un fichier .env")

from langchain_mistralai import ChatMistralAI
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
from langchain_core.prompts import ChatPromptTemplate
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings

# ===============================
# CHARGEMENT DE CHROMA
# ===============================
CHROMA_PATH = "chroma_db_COMPLETE"  

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vectorstore = Chroma(
    persist_directory=CHROMA_PATH,
    embedding_function=embeddings
)

# ===============================
# RETRIEVER - MODIFIÉ 
# ===============================
retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 10}  # 
)

# ===============================
# LLM
# ===============================
llm = ChatMistralAI(
    model="mistral-large-latest",
    temperature=0
)

# ===============================

# ===============================
prompt_template = """You are an assistant for question-answering tasks.
Use ONLY the retrieved context below to answer the question.
If the answer is not in the context, say "I don't know based on the provided documents."
Be concise and answer in maximum 3 sentences.

Context:
{context}

Question: {question}

Answer:"""

prompt = ChatPromptTemplate.from_template(prompt_template)  # 
# ===============================
# FORMATAGE
# ===============================
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

# ===============================
# RAG CHAIN
# ===============================
rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough()
    }
    | prompt
    | llm
    | StrOutputParser()
)

# ===============================
# TEST
# ===============================
response = rag_chain.invoke(
    "What is the most common pathology of the shoulder ?"
)

print(response)
```
