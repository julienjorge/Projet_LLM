```python
import os
import pdfplumber

folder = "/Users/AI/Mes Documents/VS code/projet LLM/Docugami/KG-RAG-datasets/nih-clinical-trial-protocols/data/v1/docs"

for filename in os.listdir(folder):
    if filename.lower().endswith(".pdf"):
        pdf_path = os.path.join(folder, filename)
        print(f"\n=== {filename} ===")

        with pdfplumber.open(pdf_path) as pdf:
            for page in pdf.pages:
                tables = page.extract_tables()
                if not tables:
                    continue

                for table in tables:
                    for row in table:
                        print(row)
```
